```python
from pyspark.sql import SparkSession
from pyngrok import ngrok
from google.colab import userdata
import os

ngrok.set_auth_token(userdata.get('NGORK_AUTH_TOKEN'))
public_url = ngrok.connect(4050)



```

SPARK STARTS HERE...

```python
total_cores = os.cpu_count()
optimal_partitions = total_cores * 2



spark = SparkSession.builder \
    .master("local[*]") \
    .appName("SparkLearning") \
    .config("spark.ui.port", "4050") \
    .config("spark.sql.shuffle.partitions", optimal_partitions) \
    .config("spark.memory.fraction", "0.6") \
    .config("spark.memory.storageFraction", "0.5") \
    .config("spark.executor.extraJavaOptions",
            "-verbose:gc "
            "-XX:+PrintGCDetails "
            "-XX:+PrintGCDateStamps") \
    .getOrCreate()

print("=" * 150)
print("        SPARK SESSION READY")
print("=" * 150)
print(f"  Spark UI          : {public_url}")
print(f"  Cores             : {total_cores}")
print(f"  Shuffle partitions: {optimal_partitions}")
print(f"  Memory fraction   : 0.6")
print(f"  Storage fraction  : 0.5")
print(f"  GC logging        : ON")
print("-" * 150)
print(f"  defaultParallelism: {spark.sparkContext.defaultParallelism}")
print(f"  Master            : {spark.sparkContext.master}")
print(f"  App ID            : {spark.sparkContext.applicationId}")
print("=" * 150)
```

```python
## Create the Dataframe from list of tuples

import random
import string

def random_name():
    return ''.join(random.choices(string.ascii_uppercase, k=5))

data = [(random_name(), random.randint(18, 60)) for _ in range(100000)]
print(data[:5])

# df = spark.createDataFrame(data, schema = ["name", "age"])
# here the list given is python object

# pass the datatypes in the schema
df = spark.createDataFrame(data,"name:string, age:int")
```
